# Maré de Saudade 

# 1. Amanhecer em Copacabana
Classic instrumental bossa nova, G major, 2:30-3:00  
João Gilberto soft nylon batida, Charlie Byrd delicate jazz guitar, warm upright bass, brushes, subtle vibraphone sunrise sparkle. Gentle Desafinado/Chega de Saudade vibe. Peaceful morning in Copacabana, waves in background.

# 2. A Garota Que Passa
Classic bossa nova, D♭ major, 3:00  
Female airy Portuguese vocal + Stan Getz tenor sax. The Girl from Ipanema style. Soft João guitar batida, piano, brushes. Dreamy verse-sax answers-chorus structure. Light male humming harmonies at end.

# 3. Corcovado Whispers
Slow instrumental bossa-jazz ballad, C major, 4:00  
Stan Getz silky tenor sax lead, João muted guitar, Bill Evans-style piano, upright bass, whisper brushes. Pure Quiet Nights/Corcovado emotion. Long heartfelt sax solo, misty fade.

# 4. Café com Pão de Queijo
Up-tempo happy bossa, A minor, 2:45  
Playful male Portuguese vocal, Marcos Valle crisp guitar, dancing flute, bright piano, pandeiro. Samba de Verão feel. Light scat, café ambience.

# 5. O Que Você Podia Ser
Clube da Esquina bossa ballad, E minor → G major, 3:30  
Emotional male Portuguese vocal + soft female harmonies, Milton Nascimento/Lo Borges style. Fingerstyle guitar, lush piano, crying flugelhorn. Soaring chorus, deep saudade.

# 6. Fado em Lisboa
Bossa meets fado, B minor, 3:20  
Deep female Portuguese vocal full of saudade, Amália-style. Portuguese 12-string guitar + subtle bossa rhythm, accordion touches, arco bass. Slow dramatic sunset melancholy.

# 7. Samba de Uma Nota Só
Swinging instrumental bossa-jazz, B♭ major, 3:40  
Stan Getz tenor sax + piano trading solos over classic batida guitar. One Note Samba / Manhã de Carnaval energy. Joyful head-solo-head structure.

# 8. Noite de Luar
Modern dreamy bossa-pop, A major, 3:15  
Breathy female vocal (Portuguese + some English), Bebel Gilberto/Céu style. Light acoustic + electric guitar, watery synth pads, smooth Rhodes, electronic-organic percussion. Moonlight romance.

# 9. Insensatez
Pure instrumental late-night ballad, F minor, 4:00  
Bill Evans piano trio feel + heartbreaking Stan Getz sax. How Insensitive melody. No drums, only soft percussion at end. Deep introspection.

# 10. Saudade Ficará
Grand finale bossa, D minor → D major, 4:30  
Male & female duet Portuguese, full warm band + soft choir. Reprise motifs from track 1 & 2. Nylon guitar, sax, piano, strings. Emotional lift to major, ocean fade. Eternal saudade.

# 11. Morro Não Tem Vez
Favela bossa-samba, E minor, 3:20  
Happy male Portuguese vocal + real kids shouting/singing chorus & call-response.  
Nylon guitar batida, cavaco, pandeiro, surdo, tamborim, handclaps, football shouts.  
Raw rooftop party energy, golden hour joy.
